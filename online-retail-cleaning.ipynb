# (line_total) ==> sum of every row (price*quntitey) save in new colum (line_total)

```python
# اعلي المنتجات مبيعات
# قيمه المبيعات
# قيمه العمليات السالبه
df_no_duplicates["line_total"] = df_no_duplicates["Quantity"] * df_no_duplicates["UnitPrice"]
#sort the vlaues of (line total )
df_no_duplicates.sort_values("line_total").head(5)
```

```python
df_no_duplicates = df_no_duplicates.copy()

df_no_duplicates["line_total"] = (
    df_no_duplicates["Quantity"] * df_no_duplicates["UnitPrice"]
)

df_no_duplicates[["Quantity", "UnitPrice", "line_total"]].head()
```

## (Max,Min) DATE AND TIME

```python
#minmum date and time
Min_date  = df_no_duplicates['InvoiceDate'].min()
print("Minmum date is : ",(Min_date) )
```

```python
#maxmum date and time
Max_date = df_no_duplicates['InvoiceDate'].max()
print("Maxmum date is : ",(Max_date) )
```

# Total Money


```python
# Total Money between (2010-12-01 ,2010-12-02) date is 101897.38

total_money = df_no_duplicates["line_total"].sum()

print((r"Total Money between ( 1/12/2010 and 2/12/2010 is :"),(total_money))
```

```python
#positive money
Positive_Money = df_no_duplicates.loc[
    df_no_duplicates["line_total"] > 0,
    "line_total"
].sum()

print(f"Positve money is :{Positive_Money :2f}")
```

```python
#Negative money
Negative_Money = df_no_duplicates.loc[
    df_no_duplicates["line_total"] < 0,
    "line_total"
].sum()

print(f"Negative money is :{Negative_Money :2f}")
```

```python
# check money
total = Negative_Money+ Positive_Money
print(f'total mony is : {total}')
```

#Countries Total and arrangment

```python
# total countries sorted by total money
Country_total = df_no_duplicates.groupby("Country")["line_total"].sum()
Country_total.sort_values(ascending=False)
```

```python
uk_share = (
    Country_total.loc["United Kingdom"]
    / Country_total.sum()
) * 100

print(f"UK precentage of total is : {uk_share:.2f}%")
```

# بريطانيا تتصدر صافي قيمة العمليات في العينة بنحو 97,878.97 جنيه إسترليني، وتمثل حوالي 96.06% من الإجمالي خلال 1–2 ديسمبر 2010

```python
import matplotlib.pyplot as plt

Country_total.sort_values().plot(kind="barh")

plt.title("Top  Countries — Sample Net Transaction Value")
plt.xlabel("Net transaction value (GBP)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

# Highest Prouducts


```python
product_totals = (
    df_no_duplicates.groupby("StockCode")["line_total"]
    .sum()
    .sort_values(ascending=False)
)

product_totals.head(10)
```

```python
top_codes = product_totals.head(5).index

df_no_duplicates.loc[
    df_no_duplicates["StockCode"].isin(top_codes),
    ["StockCode", "Description"]
].drop_duplicates()
```

```python
product_totals = (
    df_no_duplicates.groupby("StockCode")["line_total"]
    .sum()
    .sort_values(ascending=False)
)

product_totals.head(5).round(2)
```

```python
top_codes = product_totals.head(5).index

df_no_duplicates.loc[
    df_no_duplicates["StockCode"].isin(top_codes),
    ["StockCode", "Description"]
].drop_duplicates()
```

```python
product_totals.head(5).plot(kind="bar")

plt.title("Top 5 Product Codes — Sample Net Transaction Value")
plt.xlabel("Product code")
plt.ylabel("Net transaction value (GBP)")
plt.xticks(rotation=360)
plt.tight_layout()
plt.show()
```

# Daily Net Transaction Value

```python
daily_totals = (
    df_no_duplicates.groupby(
        df_no_duplicates["InvoiceDate"].dt.date
    )["line_total"].sum()
)

daily_totals.round(2)
```

```python
daily_totals.plot(kind="bar")

plt.title("Daily Net Transaction Value — Sample")
plt.xlabel("Date")
plt.ylabel("Net transaction value (GBP)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
